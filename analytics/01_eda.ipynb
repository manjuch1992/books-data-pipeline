# Titanic EDA

This notebook profiles the Titanic dataset and stores charts and summary tables under `analytics/charts` and `analytics/outputs`.

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

DATA_PATH = 'titanic.csv'
OUT_DIR = '.'

df = pd.read_csv(DATA_PATH)
print(df.head())
print(df.shape)
print(df.isna().sum().sort_values(ascending=False))
```

```python
missing = df.isna().sum().reset_index()
missing.columns = ['column', 'missing_values']
missing.to_csv('outputs/missing_values.csv', index=False)

survival_rates = df.groupby('sex')['survived'].mean().reset_index().rename(columns={'survived': 'survival_rate'})
survival_rates.to_csv('outputs/survival_rates.csv', index=False)

corr = df.select_dtypes(include=['number']).corr()
corr.to_csv('outputs/correlation_matrix.csv')

print('Saved summary tables to outputs/')
```

```python
os.makedirs('charts', exist_ok=True)

sns.histplot(df['age'].dropna(), bins=20, kde=True)
plt.title('Age Distribution')
plt.savefig('charts/age_histogram.png', dpi=150)
plt.close()

sns.boxplot(data=df, x='survived', y='age')
plt.title('Age by Survival')
plt.savefig('charts/age_boxplot.png', dpi=150)
plt.close()

sns.histplot(df['fare'].dropna(), bins=25, kde=True)
plt.title('Fare Distribution')
plt.savefig('charts/fare_histogram.png', dpi=150)
plt.close()

sns.boxplot(data=df, x='survived', y='fare')
plt.title('Fare by Survival')
plt.savefig('charts/fare_boxplot.png', dpi=150)
plt.close()

sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.savefig('charts/correlation_heatmap.png', dpi=150)
plt.close()

sns.barplot(data=survival_rates, x='sex', y='survival_rate')
plt.title('Survival Rate by Sex')
plt.savefig('charts/survival_by_sex.png', dpi=150)
plt.close()

survival_by_class = df.groupby('pclass')['survived'].mean().reset_index()
sns.barplot(data=survival_by_class, x='pclass', y='survived')
plt.title('Survival Rate by Passenger Class')
plt.savefig('charts/survival_by_pclass.png', dpi=150)
plt.close()

survival_by_both = df.groupby(['sex', 'pclass'])['survived'].mean().reset_index()
sns.barplot(data=survival_by_both, x='pclass', y='survived', hue='sex')
plt.title('Survival by Sex and Class')
plt.savefig('charts/survival_by_sex_pclass.png', dpi=150)
plt.close()

print('Saved charts to charts/')
```

## EDA interpretation

The initial exploration shows that sex and ticket class are highly relevant to survival patterns. Age and fare appear to differ meaningfully between survivors and non-survivors, which makes them important for the modeling phase.
